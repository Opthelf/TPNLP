# TP 1 — Du texte brut au premier modèle bigramme
Objectif : comprendre comment un corpus devient un ensemble de prédictions supervisées, puis entraîner une table qui prédit le caractère suivant.

**Livrable** : un notebook exécuté de haut en bas, deux générations — avant et après entraînement — et des réponses courtes aux questions d'interprétation.

```python
import torch
import torch.nn as nn
from torch.nn import functional as F
```

## Étape 1 : télécharger et inspecter le corpus

Relevez : la longueur du corpus ; les retours à la ligne ; la ponctuation ; les majuscules et minuscules.

```python
from urllib.request import urlretrieve
import os

url = ("https://raw.githubusercontent.com/karpathy/"
       "char-rnn/master/data/tinyshakespeare/input.txt")
urlretrieve(url, "input.txt")
print(os.path.abspath("input.txt"))

with open("input.txt", "r", encoding="utf-8") as f:
    text = f.read()

print(len(text))
print(text[:1000])
```

> **Questions d'interprétation** — Quelle est ici l'unité d'un token ? Pourquoi préciser `encoding="utf-8"` ? Le fichier est-il déjà séparé en exemples ?

**Réponses**
- **Unité d'un token** : un **caractère** (lettre, chiffre, ponctuation, espace, `\n`). Le vocabulaire contient donc 65 tokens.
- **`encoding="utf-8"`** : un fichier n'est qu'une suite d'octets ; il faut dire comment les transformer en caractères. Sans le préciser, Python utilise l'encodage par défaut du système, qui peut différer d'une machine à l'autre et mal lire (ou refuser) certains caractères.
- **Déjà séparé en exemples ?** Non : c'est un seul long texte de ~1,1 M caractères. Ce sont *nous* qui fabriquons les exemples (contexte → cible) en découpant des fenêtres dans ce texte (étapes 3 et 4).

## Étape 2 : passer des caractères aux entiers

```python
chars = sorted(list(set(text)))
vocab_size = len(chars)

stoi = {ch: i for i, ch in enumerate(chars)}
itos = {i: ch for i, ch in enumerate(chars)}

encode = lambda s: [stoi[c] for c in s]
decode = lambda ids: ''.join([ itos[i] for i in ids])

message = "hii there"
print(encode(message))
print(decode(encode(message)))
```

> **Questions d'interprétation** — Pourquoi trier `chars` ? Que garantit le test `decode(encode(message)) == message` ?

**Réponses**
- **Pourquoi trier ?** `set(text)` n'a pas d'ordre garanti : sans tri, un même caractère pourrait recevoir un identifiant différent à chaque exécution. Le tri rend la correspondance caractère ↔ entier **déterministe et reproductible**, ce qui est indispensable pour réutiliser un modèle entraîné.
- **`decode(encode(message)) == message`** garantit que le codage est **sans perte** : `stoi` et `itos` sont exactement inverses, chaque caractère a un identifiant unique et on retrouve le texte d'origine. (Valable seulement pour des caractères présents dans le corpus ; un caractère inconnu provoquerait une `KeyError`.)

## Étape 3 : fabriquer les prédictions attendues

on split le dataset en 90/10 ; 90 pour train et 10 pour val

```python
t = [1,2,3,4,5,6,7,8,9]
block_size = 8
x = t[:block_size]
y = t[1:block_size + 1]
print(x)
print(y)
```

```python
data = torch.tensor(encode(text), dtype=torch.long)

n = int(0.9 * len(data))
train_data = data[:n]
val_data = data[n:]

block_size = 8
x = train_data[:block_size]
y = train_data[1:block_size + 1]

for t in range(block_size):
    context = x[:t + 1]
    target = y[t]
    print(context, "->", target)
    
```

> **Questions d'interprétation**
> 1. Combien de prédictions produit ce bloc ?  block_size fois prédiction
> 2. Pourquoi ne pas mélanger train et validation ? pour que notre modèle n'apprennes pas par coeur les données
> 3. Pour t = 3, quels sont le contexte et la cible ? contexte = [18, 47, 56, 57] pour cible tensor(58)

**Réponses**
1. **`block_size` = 8 prédictions** : un bloc de 9 tokens donne 8 couples (contexte → cible), avec des contextes de longueur 1 à 8.
2. **Pas de mélange train/validation** : la validation doit contenir du texte **jamais vu** pendant l'entraînement. Sinon, une loss basse pourrait venir de la mémorisation, et on ne mesurerait plus la capacité à **généraliser** (on ne détecterait pas le surapprentissage).
3. **t = 3** : contexte = `[18, 47, 56, 57]` (« Firs »), cible = `58` (« t »).

## Étape 4 : échantillonner un mini-batch

```python
batch_size = 4
block_size = 8

def get_batch(split):
    source = train_data if split == "train" else val_data
    starts = torch.randint(
        len(source) - block_size, (batch_size,)
    )
    x = torch.stack([
        source[i : i + block_size] for i in starts
    ])
    y = torch.stack([
        source[i + 1 : i + block_size + 1] for i in starts
    ])
    return x, y

xb, yb = get_batch("train")
print(xb.shape, yb.shape)
```

> **Questions d'interprétation**
> 1. Pourquoi retirer `block_size` à la borne de tirage ?
> 2. Que changerait B = 32 ?
> 3. Que changerait T = 64 ?

**Réponses**
1. Chaque départ `i` doit laisser la place à `block_size` tokens pour `x` **et** à un token de plus pour `y` (décalé d'une position). L'indice le plus loin lu est `i + block_size`, qui doit exister : `i ≤ len(source) − block_size − 1`. Comme `randint` exclut sa borne haute, `len(source) − block_size` donne exactement le départ maximal valide. Sans cela, les tranches en fin de texte seraient plus courtes et `torch.stack` échouerait.
2. **B = 32** : 32 séquences indépendantes par batch au lieu de 4 → 32 × 8 = 256 prédictions par pas. Le gradient est moins bruité (moyenne sur plus d'exemples), chaque pas coûte plus de calcul/mémoire, mais le contexte de chaque prédiction ne change pas.
3. **T = 64** : des séquences de 64 caractères → plus de prédictions par séquence et des contextes jusqu'à 64 caractères. Pour le bigramme cela ne change rien à ce qu'il peut apprendre (il ne regarde que le dernier caractère), mais pour le Transformer du TP 4 cela allonge la mémoire disponible (et le coût de l'attention, en T²).

## Étape 5 : une table produit les logits

```python
class BigramLanguageModel(nn.Module):
    def __init__(self, vocab_size):
        super().__init__()
        self.token_embedding_table = nn.Embedding(
            vocab_size, vocab_size
        )

    def forward(self, idx, targets=None):
        logits = self.token_embedding_table(idx)
        # idx: (B,T) -> logits: (B, T, C)

        if targets is None:
            loss = None
        else:
            B, T, C = logits.shape
            logits = logits.view(B*T, C)
            targets = targets.view(B*T)
            loss = F.cross_entropy(logits, targets)
        return logits, loss
```

> **Questions d'interprétation** — Pourquoi C × C ici ? Où retrouvera-t-on d_emb et la couche `lm_head` dans le TP 4 ?

**Réponses**
- **C × C** : une ligne par caractère **courant** (l'entrée choisit la ligne) et une colonne par caractère **suivant possible** (les scores/logits). Entrée et sortie appartiennent au même vocabulaire, donc la table est `vocab_size × vocab_size` (65 × 65). La case (a, b) est le score de « b vient après a ».
- **Dans le TP 4**, cette table est séparée en deux :
  - `token_embedding_table = nn.Embedding(vocab_size, n_embd)` : chaque caractère devient un vecteur de dimension **d_emb** (64) ;
  - `lm_head = nn.Linear(n_embd, vocab_size)` : ramène ce vecteur vers les 65 logits.

  Entre les deux s'insèrent les blocs (attention + MLP).

## Étapes 6 et 7 : générer, puis apprendre
Générez une première fois avec les poids aléatoires, entraînez la table, puis relancez exactement la même procédure pour observer ce qui a changé.

### Étape 6 — génération (méthode de `BigramLanguageModel`)

```python
def generate(self, idx, max_new_tokens):
    for _ in range(max_new_tokens):
        logits, _ = self(idx)
        logits = logits[:, -1, :]
        probs = F.softmax(logits, dim=-1)
        idx_next = torch.multinomial(
            probs, num_samples=1
        )
        idx = torch.cat((idx, idx_next), dim=1)
    return idx

BigramLanguageModel.generate = generate
```

```python
# Génération avant entraînement (poids aléatoires)
model = BigramLanguageModel(vocab_size)
idx = torch.zeros((1, 1), dtype=torch.long)
print(decode(model.generate(idx, max_new_tokens=100)[0].tolist()))
```

### Étape 7 — entraînement

```python
optimizer = torch.optim.AdamW(
    model.parameters(), lr=1e-3
)

for step in range(1000):
    xb, yb = get_batch("train")
    logits, loss = model(xb, yb)
    optimizer.zero_grad(set_to_none=True)
    loss.backward()
    optimizer.step()

print(loss.item())
```

```python
# Génération après entraînement
print(decode(model.generate(idx, max_new_tokens=100)[0].tolist()))
```

> **Bilan du TP 1** — Pourquoi ne garde-t-on que le dernier pas de temps pendant la génération ? Pourquoi aplatit-on B × T pendant la loss ?
> Comparez la loss et la structure du texte avant/après 100 étapes, puis avec 1 000 étapes.

**Réponses**
- **Dernier pas de temps** : `logits` contient une prédiction pour chaque position, mais les positions `t < T−1` prédisent des tokens **déjà connus**. Seule la dernière position prédit le token **inconnu** à générer. `logits[:, -1, :]` passe de `(B, T, C)` à `(B, C)`, la forme attendue par `softmax` puis `multinomial`.
- **Aplatir B × T** : `F.cross_entropy` attend des logits `(N, C)` et des cibles `(N,)`. Chaque position de chaque séquence est une prédiction indépendante ; on les met donc toutes « à la suite » (N = B·T) pour calculer la loss moyenne en une fois.
- **Avant / après** : une prédiction uniforme donnerait une loss de `ln(65) ≈ 4,17`. Au départ, la loss est en général un peu **au-dessus** (≈ 4,5–4,9), car la table est initialisée avec des valeurs aléatoires N(0, 1) : le modèle n'est pas uniforme mais « confiant au hasard ». Le texte généré est alors une suite aléatoire de caractères.
- Ici, après 1 000 étapes (`lr = 1e-3`, `batch_size = 4`), la loss affichée est **3,86** : elle passe sous ln(65), donc le modèle fait déjà mieux qu'un tirage uniforme. La baisse reste modeste : chaque étape ne voit que 4 × 8 = 32 prédictions, et 1 000 petites mises à jour ne suffisent pas à bien remplir une table de 65 × 65 = 4 225 scores. Le texte reste donc très proche de l'aléatoire, avec un peu plus d'espaces et de retours à la ligne. Il faut beaucoup plus d'étapes (≈ 10 000) ou un batch plus grand pour voir apparaître des paires de lettres plausibles.
- Avec seulement 100 étapes, la loss bouge à peine par rapport au départ. Remarque : `loss.item()` est la loss d'**un seul** mini-batch, donc une valeur bruitée.
